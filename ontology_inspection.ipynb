# Ontology Inspection

This notebook parses `ontology.ttl` and lists ontology entities:
- ontology metadata
- classes
- object properties
- datatype properties
- annotation properties

```python
from pathlib import Path
from rdflib import Graph
from rdflib.namespace import RDF, RDFS, OWL
import pandas as pd
```

```python
ontology_path = Path('ontology.ttl').resolve()
print('Ontology file:', ontology_path)
print('Exists:', ontology_path.exists())

g = Graph()
g.parse(ontology_path, format='turtle')
print('Triples loaded:', len(g))
```

```python
def local_name(uri):
    s = str(uri)
    if '#' in s:
        return s.rsplit('#', 1)[1]
    return s.rsplit('/', 1)[-1]

def get_label(node):
    val = g.value(node, RDFS.label)
    return str(val) if val else ''

def get_comment(node):
    val = g.value(node, RDFS.comment)
    return str(val) if val else ''

def rows_for(rdf_type):
    rows = []
    for node in sorted(set(g.subjects(RDF.type, rdf_type)), key=lambda x: str(x)):
        rows.append({
            'uri': str(node),
            'local_name': local_name(node),
            'label': get_label(node),
            'comment': get_comment(node),
        })
    return pd.DataFrame(rows)
```

```python
ontology_df = rows_for(OWL.Ontology)
classes_df = rows_for(OWL.Class)
object_props_df = rows_for(OWL.ObjectProperty)
datatype_props_df = rows_for(OWL.DatatypeProperty)
annotation_props_df = rows_for(OWL.AnnotationProperty)

print('Ontology nodes:', len(ontology_df))
print('Classes:', len(classes_df))
print('Object properties:', len(object_props_df))
print('Datatype properties:', len(datatype_props_df))
print('Annotation properties:', len(annotation_props_df))
```
