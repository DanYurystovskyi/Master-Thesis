## Combine CDF match data into one JSON file
Going through the `cdf_batch_output` folder, reads all match folders
to combine their JSON files into a single file.

```python
from pathlib import Path
import json

project_root = Path(".").resolve()
cdf_dir = project_root / "cdf_batch_output"
output_path = project_root / "cdf_all_matches.json"
```

```python
# We take a match id as the level 0 in JSON and nest all the data inside of it.
def load_match_folder(match_dir: Path) -> dict:
    match_id = match_dir.name
    data = {"match_id": match_id}

    meta_path = match_dir / "match_meta_cdf.json"
    sheet_path = match_dir / "match_sheet_cdf.json"
    event_path = match_dir / "event_cdf.json"

    # Match meta
    if meta_path.exists():
        with open(meta_path, "r", encoding="utf-8") as f:
            data["meta"] = json.load(f)
    else:
        data["meta"] = None

    # Match sheet
    if sheet_path.exists():
        with open(sheet_path, "r", encoding="utf-8") as f:
            data["sheet"] = json.load(f)
    else:
        data["sheet"] = None

    # Events
    if event_path.exists():
        with open(event_path, "r", encoding="utf-8") as f:
            data["events"] = json.load(f)
    else:
        data["events"] = None

    return data
```

```python
# Find all match folders inside of the target location
match_dirs = sorted([p for p in cdf_dir.iterdir() if p.is_dir()])
all_matches = []

for match_dir in match_dirs:
    print("Loading match:", match_dir.name)
    match_data = load_match_folder(match_dir)
    all_matches.append(match_data)

# and combining those together into a single JSON file
with open(output_path, "w", encoding="utf-8") as f:
    json.dump(all_matches, f, ensure_ascii=False, indent=2)
```
